```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
#capa de datos
df = pd.read_csv("netflix_titles.csv")
```

```python
# Mostrar primeras 5 filas
df.head()
```

```python
print("Dimensiones del dataset:")
print(df.shape)
```

```python
print("Columnas del dataset:")
print(df.columns)
```

```python
print("Tipos de datos:")
print(df.dtypes)
```

```python
#informacion general 
df.info()
```

```python
#capa de informacion
#valores nulos 
print(df.isnull().sum())
```

```python
porcentaje_nulos = (df.isnull().sum() / len(df)) * 100

print(porcentaje_nulos)
```

```python
print("Cantidad de duplicados:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
#eliminar una columna 
df.drop(columns=["director"], inplace=True)
```

```python
df.columns 
```

```python
print(df.isnull().sum())
```

```python
#Rellenar valores nulos
df.fillna("Desconocido", inplace=True)
```

```python
#grafica de pelicula vs tv show 
plt.figure(figsize=(6,4))

sns.countplot(x='type', data=df)

plt.title("Cantidad de Películas y Series en Netflix")
plt.xlabel("Tipo")
plt.ylabel("Cantidad")

plt.show()
```

```python
#grafica de paises con mas contenido 
top_paises = df['country'].value_counts().head(10)

plt.figure(figsize=(10,5))

sns.barplot(x=top_paises.values, y=top_paises.index)

plt.title("Top 10 países con más contenido")
plt.xlabel("Cantidad")
plt.ylabel("País")

plt.show()
```

```python
clasificacion de contenido 
plt.figure(figsize=(10,5))

sns.countplot(
    y='rating',
    data=df,
    order=df['rating'].value_counts().index
)

plt.title("Clasificación del contenido")

plt.show()
```

```python
plt.figure(figsize=(12,5))

sns.histplot(df['release_year'], bins=20)

plt.title("Contenido por año")
plt.xlabel("Año")
plt.ylabel("Cantidad")

plt.show()
```
