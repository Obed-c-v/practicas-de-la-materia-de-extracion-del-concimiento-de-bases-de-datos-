```python
#limpieza 
#nombres de columnas en minusculas +
df.columns= df.columns.str.strip().str.lower()
```

```python
df.columns 
```

```python
df.isnull().sum()
```

```python
#eliminar una columna 
df.drop(columns=["cabin"], inplace=True)
```

```python
df.columns 
```

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
df.isnull().sum()
```

```python
#rrellena la columna fare con el valor mas repetido (moda)
df["fare"]=df["fare"].fillna(df["fare"].mode()[0])
```

```python
#total de pasagero por clase
plt.figure(figsize=(10,5)) #Tamaño
sns.countplot(df, x="pclass")
plt.title("total de pasajeros por clase")
plt.xlabel("Clase")
plt.show()
```

```python
#Graficamos las clases por genero 
plt.figure(figsize=(8,4))
sns.countplot(df,x="sex", hue="pclass")
plt.title("Total de genero por clases")
plt.xlabel("Genero")
plt.ylabel("Total")
plt.show()
```

```python
#Mostramos la edad por genero y clase 

plt.figure(figsize=(6,4))
sns.violinplot(df,x="sex", y="age", hue="pclass")
plt.title("rango de edad por genero y clases")
plt.xlabel("Genero")
plt.ylabel("edad")
plt.show()
```

```python
#los pasajeros de primera clase pagan tarifas mas altas 
df.groupby("pclass")["fare"].mean()
```
