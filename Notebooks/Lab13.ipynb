```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub

# Importaciones específicas de Scikit-Learn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
from sklearn.metrics import (accuracy_score, confusion_matrix, classification_report, ConfusionMatrixDisplay)
```

```python
from sklearn.datasets import load_iris 
iris = load_iris(as_frame = True)
```

```python
print(iris.DESCR)
```

```python
dfIris = iris.frame 
```

```python
###renombrar la columna target

dfIris.rename(columns={"target": "especies"}, inplace = True)
```

```python
dfIris["especies"] = dfIris ["especies"].map({
    0: "setosa",
    1:"versicolor",
    2:"virginica"
})
```

```python
dfIris.head()
```

```python
dfIris.tail()
```

## identificar variables predictoras y variable objetivo 

```python
##variables predictoras 

X = dfIris.drop(columns= ["especies"])
```

```python
print("variables predictoras")
print(X.columns)
```

```python
# variable objetivo 
y = dfIris["especies"]
```

```python
print("variable objetivo")
print(y.name)
```

## dividir los datos en prueba y entrenamiento 

vamos a distribuir los datos en 80% para entrenamiento y 20 % para prueba 

```python
X_train, X_test , y_train, y_tests= train_test_split(
    X,y, test_size= 0.20,random_state=42, stratify= y 
)
```

```python
print("x_train",X_train.shape)
print("X_tests",X_test.shape)
print("y_train",y_train.shape)
print("y_tests",y_tests.shape)
```

## estacaldo de variables 

```python
scaler = StandardScaler()
```

```python
# escalamos las variables 
X_train_escalado= scaler.fit_transform(X_train)
X_test_escalado= scaler.transform(X_test)
```

```python
print ("sin escalar")
print (X_train.head())
print("escalado")
print(X_train_escalado[:5])
```

## entrenamiento del modelo KNN 
en esta prueba utilizaremos k=3 

```python
modelo =KNeighborsClassifier(n_neighbors=3)
```

```python
modelo.fit(
    X_train_escalado,
    y_train
)
```

## Realizando  predicciones  con las variables de prueba 

```python
prediccion = modelo .predict(X_test_escalado)
```

```python
print(prediccion[:10])
```

```python
# comparacion entre valores reales y valores predecidos 

comparacion = pd.DataFrame({
    "especie real": y_tests.values, 
    "especie predicha" : prediccion
}) 
```

```python
comparacion.head(30)
```

## Evaluacion del modelo 

```python
accuracy = accuracy_score(y_tests, prediccion)
print(f"accuracy/exactitud: {accuracy:.4f}")
```

```python
## matriz de confusion 
cm = confusion_matrix(y_tests, prediccion)
```

```python
cm_df = pd.DataFrame(
    cm,
    index= modelo.classes_,
    columns = modelo.classes_

)

cm_df
```

##  Clasification report
proporciona metriscas que permite evaluar el desempeño de cada clase: precision, recall,f1-score, support 

```python
print(classification_report(y_tests,prediccion))
```
