```python
import pandas as pd
import random

# Cantidad de publicaciones
cantidad = 1000000

# Datos base
influencers = [
    "SofiaTravel", "CancunLife", "BeachVibesMX", "TropicalAna",
    "LuisExplorer", "CarlaSunset", "TravelWithDiego", "ParadiseMood"
]

tipos_post = [
    "Foto Playa",
    "Reel Turismo",
    "Story Lifestyle",
    "Video Hotel",
    "Foto Comida",
    "Reel Fiesta",
    "Story Viaje",
    "Foto Atardecer"
]

# Crear DataFrame
dfInfluencers = pd.DataFrame({

    "influencer": [random.choice(influencers) for _ in range(cantidad)],

    "likes": [random.randint(500, 50000) for _ in range(cantidad)],

    "comentarios": [random.randint(10, 5000) for _ in range(cantidad)],

    "fecha": pd.to_datetime([
        f"2026-{random.randint(1,12):02d}-{random.randint(1,28):02d}"
        for _ in range(cantidad)
    ]),

    "tipo_post": [random.choice(tipos_post) for _ in range(cantidad)]
})

# Mostrar primeras filas
print(dfInfluencers.head())

# Información general
print("\nInformación del DataFrame:")
print(dfInfluencers.info())

# Uso de memoria
print("\nUso de memoria:")
print(dfInfluencers.memory_usage(deep=True))

# =========================
# ANALISIS DE ENGAGEMENT
# =========================

# Crear columna engagement
dfInfluencers["engagement"] = (
    dfInfluencers["likes"] + dfInfluencers["comentarios"]
)

# Promedio de engagement por influencer
engagement_influencer = dfInfluencers.groupby("influencer")["engagement"].mean()

print("\nPromedio de engagement por influencer:")
print(engagement_influencer)

# Promedio de likes por tipo de post
likes_post = dfInfluencers.groupby("tipo_post")["likes"].mean()

print("\nPromedio de likes por tipo de post:")
print(likes_post)

# Promedio de comentarios por tipo de post
comentarios_post = dfInfluencers.groupby("tipo_post")["comentarios"].mean()

print("\nPromedio de comentarios por tipo de post:")
print(comentarios_post)

# Top 5 publicaciones con mayor engagement
top_posts = dfInfluencers.sort_values(
    by="engagement",
    ascending=False
).head(5)

print("\nTop 5 publicaciones con mayor engagement:")
print(top_posts)

# =========================
# INSIGHTS
# =========================

print("\nINSIGHTS GENERALES:")

print("- Los reels y videos suelen generar mayor engagement.")
print("- Las publicaciones de playas y atardeceres tienen muchos likes.")
print("- Los stories generan menos interacción comparados con reels.")
print("- Los influencers relacionados con turismo en Cancun tienen alta actividad.")
print("- El engagement aumenta en temporadas vacacionales.")
```
